```python
#alpha property
# alpha value lies between the 0.0 - 1.0
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(1,21)
y = x**2
plt.plot(x,y,'s-.k', alpha= 1.0) #alpha = 0.0 wont give a line but 1.0 gives a line
plt.title("line plot")
plt.xlabel("values")
plt.ylabel("square of x")
```

```python
# line width shows the width of the line(ls)
# marker size shows the size of the marker (ms)
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(1,21)
y = x**2
plt.plot(x,y, 'h-g', alpha = 1.0, lw=15, ms=20)
plt.title("line plot")
plt.xlabel("values")
plt.ylabel("square of x")
```

```python
import numpy as np
import matplotlib.pyplot as plt
names = ['dharani','sahana','keerthi','hema','shravani']
eng_marks = [95,90,85,80,75]
math_marks = [75,80,70,65,55]
total_marks = eng_marks+math_marks
plt.bar(names,eng_marks,color = 'r',label = 'English')
plt.bar(names,math_marks,bottom= eng_marks,color = 'g', label = 'maths')
for i in range(len(names)):
    plt.text(names[i],(eng_marks[i]/2),str(eng_marks[i]),ha='center',color='white',weight=1000)
    plt.text(names[i],(eng_marks[i]+math_marks[i]/2),str(math_marks[i]),ha='center',color='white',weight=1000)
    plt.text(names[i],(eng_marks[i]+2),str(total_marks[i]),ha='center',color='black',weight=1000)
plt.xlabel("names of the student",color = "magenta",fontsize = 15, weight=1000)
plt.ylabel("marks",color='#570b66',fontsize=15,weight=1000)
plt.title("student-wise marks",color = '#ED0A3F',fontsize=15,weight=1000)
plt.show()
```
